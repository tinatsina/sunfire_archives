```python
import json

with open("./en_US/runesReforged.json", mode="r", encoding="utf-8") as file:
    data = json.load(file)
```

```python
rune_list = []

for x in range(5):
    for y in data[x]['slots'][0]['runes']:
        rune_list.append((y['key'], y['id']))

print(rune_list)
```

```python

for x in range(5):
    rune_list.append((data[x]['name'], x))


rune_list.reverse()
```

```python
with open("rune_dictionary.json", mode="w") as file:
    json.dump(dict(rune_list) ,file, indent=4)
```
