```python
import json
import os
import pandas as pd

df = pd.read_csv("./HexGate/Game_Log_00000.csv").dropna()
```

```python
with open("./summoners_dictionary.json", mode="r", encoding="utf-8") as file:
    summoner_dict = json.load(file)

with open("./item_dictionary.json", mode="r", encoding="utf-8") as file:
    item_dict = json.load(file)

with open("./rune_dictionary.json", mode="r", encoding="utf-8") as file:
    rune_dict = json.load(file)

with open("./champion_dictionary.json", mode="r", encoding="utf-8") as file:
    champ_dict = json.load(file)

df = df.replace(summoner_dict)
df = df.replace(item_dict)
df = df.replace(rune_dict)
df = df.replace(champ_dict)
```

```python
champions = df[[
    'BLUE_TEAM_TOP',
    'BLUE_TEAM_JNG',
    'BLUE_TEAM_MID',
    'BLUE_TEAM_ADC',
    'BLUE_TEAM_SUP',
    'RED_TEAM_TOP',
    'RED_TEAM_JNG',
    'RED_TEAM_MID',
    'RED_TEAM_ADC',
    'RED_TEAM_SUP'
]]
```

```python
unique_list = pd.unique(
    champions.values.ravel()
)

print(len(unique_list))
```
