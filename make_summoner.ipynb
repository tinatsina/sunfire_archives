```python
import json

with open("./en_US/summoner.json", mode="r", encoding="utf-8") as file:
    data = json.load(file)
```

```python
count = 0

summoners = []

for summoner in data['data']:
    summoners.append((data['data'][summoner]['name'], count))
    count +=1

print(summoners)

summoners_dict = dict(summoners)

with open("summoners_dictionary.json", mode="w") as file:
    json.dump(summoners_dict, file, indent=4)
```
